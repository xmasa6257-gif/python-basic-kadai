```python
class Human:
    """名前と年齢を持つ人間を表すクラス。"""

    def __init__(self, name, age):
        """コンストラクタ。属性に値を代入する。

        Args:
            name: 名前
            age: 年齢
        """
        self.name = name
        self.age = age

    def printinfo(self):
        """名前と年齢を標準出力する。"""
        print("名前:", self.name)
        print("年齢:", self.age)


# インスタンスを変数に代入する
human1 = Human("侍 太郎", 25)
human2 = Human("侍 花子", 30)

# メソッドを使用して名前と年齢を出力する
human1.printinfo()
human2.printinfo()
```
