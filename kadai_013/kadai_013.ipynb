```python
def calc_price_with_tax(price, tax_rate):
    """商品の金額に消費税を加えた金額を返す。

    Args:
        price: 商品の金額
        tax_rate: 消費税率（％）

    Returns:
        消費税を加えた金額
    """
    return price + price * tax_rate / 100


# 第1引数に商品の金額「110」、第2引数に消費税（10%）の「10」を設定
result = calc_price_with_tax(110, 10)
print(result)
```
