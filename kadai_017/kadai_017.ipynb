```python
# グローバルスコープの変数（成人とみなす年齢）
ADULT_AGE = 20


class Human:
    """名前と年齢の属性を持つ人間を表すクラス。"""

    def __init__(self, name, age):
        """コンストラクタ。属性に値を代入する。"""
        self.name = name
        self.age = age

    def check_adult(self):
        """年齢が20以上かどうかを判定して標準出力する。"""
        if self.age >= ADULT_AGE:
            print(f"{self.name}さん（{self.age}歳）は大人です。")
        else:
            print(f"{self.name}さん（{self.age}歳）は大人ではありません。")


def create_humans(human_data):
    """ディクショナリのリストから Human インスタンスのリストを作成して返す。

    Args:
        human_data (list): 名前と年齢を持つディクショナリのリスト

    Returns:
        list: Human インスタンスのリスト
    """
    # ローカルスコープの変数（この関数の中だけで有効）
    humans = []
    for data in human_data:
        humans.append(Human(data["name"], data["age"]))
    return humans


# ディクショナリを要素に持つリスト
human_data = [
    {"name": "侍 太郎", "age": 25},
    {"name": "侍 花子", "age": 18},
    {"name": "侍 次郎", "age": 20},
    {"name": "侍 三郎", "age": 15},
]

# 関数を呼び出してインスタンスのリストを作成する
human_list = create_humans(human_data)

# リストの要素数分だけ check_adult メソッドを呼び出す
for human in human_list:
    human.check_adult()
```
